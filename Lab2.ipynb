```python
import cv2, time
import numpy as np
import matplotlib.pyplot as plt

img  = cv2.imread('/content/photo.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float64)
```

```python
plt.figure(figsize=(6, 6))
plt.imshow(gray, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
plt.show()
```

```python
# adds random intensity variations to pixels
def add_gaussian_noise(g, sigma=15):
    return np.clip(g + np.random.normal(0, sigma, g.shape), 0, 255)

# randomly changes some pixels to black (0) or white (255).
def add_salt_pepper(g, p=0.05):
    out = g.copy();
    r = np.random.rand(*g.shape)
    out[r < p/2] = 0;
    out[r > 1 - p/2] = 255
    return out
```

```python
def my_convolve(image, kernel):
    kh, kw = kernel.shape
    ph, pw = kh // 2, kw // 2
    padded = np.pad(image, ((ph, ph), (pw, pw)))
    out = np.zeros_like(image)
    k = np.flip(kernel)              # true convolution flips the kernel
    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            region = padded[i:i+kh, j:j+kw]
            out[i, j] = np.sum(region * k)   # multiply-and-sum
    return out
```

```python
#Task 1
#1
box_kernel = np.ones((3, 3), dtype=np.float64) / 9
small = cv2.resize(gray, (256, 256))
my_result = my_convolve(small, box_kernel)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(small, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(my_result, cmap='gray')
plt.title("My 3x3 Box Filter")
plt.axis('off')

plt.show()
```

```python
#2
cv_result = cv2.filter2D(
    small,
    ddepth=-1, # keep the same data type as the input image
    kernel=box_kernel,
    borderType=cv2.BORDER_CONSTANT  # padding with zero
)

diff = np.abs(my_result - cv_result)

print("Maximum absolute difference:")
print(np.max(diff)) # the worst mismatch between the two implementations

#3
# The difference is not always zero because our function performs true convolution by flipping the kernel, whereas cv2.filter2D() performs correlation without flipping it. For symmetric kernels such as the box filter, flipping does not change the kernel, so both operations produce mathematically identical results. However, small floating-point rounding differences may still occur. For non-symmetric kernels, convolution and correlation can produce significantly different results.
```

```python
# Task 2
#1
box9 = cv2.blur( gray, (9, 9) )
gauss1 = cv2.GaussianBlur(gray, (0, 0), 1)
gauss3 = cv2.GaussianBlur(gray, (0, 0), 3)
gauss7 = cv2.GaussianBlur(gray, (0, 0), 7)

plt.figure(figsize=(15, 8))

images = [
    gray,
    box9,
    gauss1,
    gauss3,
    gauss7
]

titles = [
    "Original",
    "9x9 Box",
    "Gaussian σ=1",
    "Gaussian σ=3",
    "Gaussian σ=7"
]

for i in range(len(images)):

    plt.subplot(2, 3, i + 1)

    plt.imshow(images[i], cmap='gray')

    plt.title(titles[i])

    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
#2
h, w = gray.shape

y1 = h // 4
y2 = 3 * h // 4

x1 = w // 4
x2 = 3 * w // 4

plt.figure(figsize=(15, 4))

for i in range(len(images)):

    crop = images[i][y1:y2, x1:x2]

    plt.subplot(1, 5, i + 1)

    plt.imshow(crop, cmap='gray')

    plt.title(titles[i])

    plt.axis('off')

plt.tight_layout()
plt.show()
#3
#The box filter gives equal weight to all pixels inside the neighborhood, which can make edges look more uniformly smeared. The Gaussian filter gives higher weights to pixels near the center and lower weights to distant pixels, producing a smoother and more natural transition near edges. Increasing σ increases the amount of blur.
```

```python
#Task 3
#1
plt.figure(figsize=(15, 4))

plt.subplot(1, 4, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original")
plt.axis('off')
blur  = cv2.GaussianBlur(gray, (0, 0), 3)
for i,a in enumerate([0.5, 1.0, 2.0]):
    sharp = np.clip(gray + a * (gray - blur), 0, 255)
    plt.subplot(1, 4, i + 2)
    plt.imshow(sharp, cmap='gray')
    plt.title(f"alpha = {a}")
    plt.axis('off')
    # display each result with its alpha
# sharp=original+α(original−blur)


plt.tight_layout()
plt.show()

#2
# At α = 2, the image may become over-sharpened. Strong halos or ringing can appear around edges, and noise may also become more visible. This happens because a large α strongly amplifies the high-frequency components contained in gray - blur.
```

```python
#Task 4
#1
gaussian_noisy = add_gaussian_noise( gray,sigma=15)

sp_noisy = add_salt_pepper(gray,p=0.05)

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(gray, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(gaussian_noisy, cmap='gray')
plt.title("Gaussian Noise")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(sp_noisy, cmap='gray')
plt.title("Salt & Pepper Noise")
plt.axis('off')

plt.show()
```

```python
#2
# a Gaussian Blur
gaussian_gauss = cv2.GaussianBlur( gaussian_noisy,(0, 0),2)

sp_gauss = cv2.GaussianBlur( sp_noisy, (0, 0),  2)

#b Median Filter
gaussian_uint8 = gaussian_noisy.astype(np.uint8)
sp_uint8 = sp_noisy.astype(np.uint8)

gaussian_median = cv2.medianBlur(
    gaussian_uint8,
    5
) #kernel size=5 needs the image to be unit8

sp_median = cv2.medianBlur(
    sp_uint8,
    5
)
```

```python
#3
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.imshow(gaussian_gauss, cmap='gray')
plt.title("Gaussian Noise + Gaussian Filter")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(gaussian_median, cmap='gray')
plt.title("Gaussian Noise + Median Filter")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(sp_gauss, cmap='gray')
plt.title("Salt & Pepper + Gaussian Filter")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(sp_median, cmap='gray')
plt.title("Salt & Pepper + Median Filter")
plt.axis('off')

plt.tight_layout()
plt.show()

#4
# Salt-and-pepper noise consists of extreme outlier values, usually 0 and 255. A median filter sorts the neighborhood values and selects the middle value, so isolated extreme pixels have little effect on the output. In contrast, averaging-based filters include the corrupted pixel in the calculation, causing its intensity to spread to neighboring pixels. Therefore, median filtering is much more effective for salt-and-pepper noise.
```
